```python
# ============================================================
# MARKET BASKET ANALYSIS
# 1. APRIORI
# 2. FP-GROWTH
# 3. ECLAT
# 4. PCA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import (
    apriori,
    fpgrowth,
    association_rules
)

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler


# ============================================================
# STEP 1: LOAD DATASET
# ============================================================

file_path = "Market_Basket_Optimisation.csv"

df = pd.read_csv(
    file_path,
    header=None
)

print("Dataset Shape:", df.shape)
print("\nFirst 5 Transactions:")
print(df.head())


# ============================================================
# STEP 2: CONVERT DATA INTO TRANSACTIONS
# ============================================================

transactions = []

for row in df.values:

    transaction = [
        str(item).strip()
        for item in row
        if pd.notna(item)
        and str(item).strip() != ""
    ]

    transactions.append(transaction)

print("\nNumber of Transactions:",
      len(transactions))


# ============================================================
# STEP 3: ONE-HOT ENCODING
# ============================================================

te = TransactionEncoder()

te_array = te.fit(
    transactions
).transform(
    transactions
)

basket = pd.DataFrame(
    te_array,
    columns=te.columns_
)

print("\nOne-Hot Encoded Data:")
print(basket.head())

print("\nNumber of Unique Items:",
      len(basket.columns))


# ============================================================
# 1. APRIORI ALGORITHM
# ============================================================

print("\n")
print("=" * 60)
print("1. APRIORI ALGORITHM")
print("=" * 60)

# Minimum support
MIN_SUPPORT = 0.01

frequent_itemsets_apriori = apriori(
    basket,
    min_support=MIN_SUPPORT,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(
    frequent_itemsets_apriori
    .sort_values(
        "support",
        ascending=False
    )
    .head(20)
)


# Generate association rules
rules_apriori = association_rules(
    frequent_itemsets_apriori,
    metric="confidence",
    min_threshold=0.30
)

# Sort by lift
rules_apriori = rules_apriori.sort_values(
    by="lift",
    ascending=False
)

print("\nTop Apriori Association Rules:")

print(
    rules_apriori[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(20)
)


# Save Apriori results
frequent_itemsets_apriori.to_csv(
    "apriori_frequent_itemsets.csv",
    index=False
)

rules_apriori.to_csv(
    "apriori_rules.csv",
    index=False
)


# ============================================================
# 2. FP-GROWTH ALGORITHM
# ============================================================

print("\n")
print("=" * 60)
print("2. FP-GROWTH ALGORITHM")
print("=" * 60)

frequent_itemsets_fp = fpgrowth(
    basket,
    min_support=MIN_SUPPORT,
    use_colnames=True
)

print("\nFrequent Itemsets:")

print(
    frequent_itemsets_fp
    .sort_values(
        "support",
        ascending=False
    )
    .head(20)
)


# Generate rules
rules_fp = association_rules(
    frequent_itemsets_fp,
    metric="confidence",
    min_threshold=0.30
)

rules_fp = rules_fp.sort_values(
    by="lift",
    ascending=False
)

print("\nTop FP-Growth Association Rules:")

print(
    rules_fp[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(20)
)


# Save FP-Growth results
frequent_itemsets_fp.to_csv(
    "fpgrowth_frequent_itemsets.csv",
    index=False
)

rules_fp.to_csv(
    "fpgrowth_rules.csv",
    index=False
)


# ============================================================
# 3. ECLAT ALGORITHM
# ============================================================

print("\n")
print("=" * 60)
print("3. ECLAT ALGORITHM")
print("=" * 60)


# ------------------------------------------------------------
# Create vertical transaction database
# ------------------------------------------------------------

item_transaction_ids = {}

for transaction_id, transaction in enumerate(
    transactions
):

    for item in set(transaction):

        if item not in item_transaction_ids:

            item_transaction_ids[item] = set()

        item_transaction_ids[item].add(
            transaction_id
        )


# ------------------------------------------------------------
# ECLAT recursive function
# ------------------------------------------------------------

def eclat(
    prefix,
    items,
    min_support_count,
    total_transactions,
    results
):

    while items:

        item, item_tidset = items.pop()

        support_count = len(
            item_tidset
        )

        if support_count >= min_support_count:

            new_itemset = (
                prefix + [item]
            )

            support = (
                support_count
                / total_transactions
            )

            results.append(
                (
                    new_itemset,
                    support
                )
            )

            suffix = []

            for other_item, other_tidset in items:

                intersection = (
                    item_tidset
                    & other_tidset
                )

                if len(intersection) >= min_support_count:

                    suffix.append(
                        (
                            other_item,
                            intersection
                        )
                    )

            eclat(
                new_itemset,
                suffix,
                min_support_count,
                total_transactions,
                results
            )


# Minimum support count
min_support_count = int(
    np.ceil(
        MIN_SUPPORT
        * len(transactions)
    )
)

# Prepare items
items = sorted(
    item_transaction_ids.items(),
    key=lambda x: len(x[1])
)

eclat_results = []

eclat(
    [],
    items,
    min_support_count,
    len(transactions),
    eclat_results
)


# ------------------------------------------------------------
# Convert ECLAT results to DataFrame
# ------------------------------------------------------------

eclat_df = pd.DataFrame(
    eclat_results,
    columns=[
        "itemsets",
        "support"
    ]
)

eclat_df["itemsets"] = (
    eclat_df["itemsets"]
    .apply(
        lambda x: ", ".join(x)
    )
)

eclat_df = eclat_df.sort_values(
    by="support",
    ascending=False
)

print("\nTop ECLAT Frequent Itemsets:")

print(
    eclat_df.head(20)
)


# Save ECLAT results
eclat_df.to_csv(
    "eclat_frequent_itemsets.csv",
    index=False
)


# ============================================================
# 4. PCA
# ============================================================

print("\n")
print("=" * 60)
print("4. PCA - PRINCIPAL COMPONENT ANALYSIS")
print("=" * 60)


# Convert Boolean values to numerical values
X = basket.astype(float)


# Standardize data
scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X
)


# Apply PCA
pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_scaled
)


# Create PCA DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=[
        "PC1",
        "PC2"
    ]
)

print("\nPCA Result:")
print(
    pca_df.head()
)


# Explained variance
explained_variance = (
    pca.explained_variance_ratio_
)

print("\nExplained Variance:")

print(
    "PC1:",
    round(
        explained_variance[0] * 100,
        2
    ),
    "%"
)

print(
    "PC2:",
    round(
        explained_variance[1] * 100,
        2
    ),
    "%"
)

print(
    "Total Explained Variance:",
    round(
        explained_variance.sum() * 100,
        2
    ),
    "%"
)


# Save PCA results
pca_df.to_csv(
    "pca_results.csv",
    index=False
)


# ============================================================
# PCA GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    s=10,
    alpha=0.5
)

plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.title(
    "PCA Visualization of Market Basket Transactions"
)

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    "pca_visualization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# TOP FREQUENT ITEMS GRAPH
# ============================================================

item_support = basket.mean().sort_values(
    ascending=False
).head(15)

plt.figure(
    figsize=(12, 6)
)

item_support.sort_values().plot(
    kind="barh"
)

plt.xlabel(
    "Support"
)

plt.ylabel(
    "Items"
)

plt.title(
    "Top 15 Most Frequent Items"
)

plt.tight_layout()

plt.savefig(
    "top_15_items.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# COMPARISON OF FREQUENT ITEMSETS
# ============================================================

print("\n")
print("=" * 60)
print("ALGORITHM COMPARISON")
print("=" * 60)

print(
    "\nApriori Frequent Itemsets:",
    len(frequent_itemsets_apriori)
)

print(
    "FP-Growth Frequent Itemsets:",
    len(frequent_itemsets_fp)
)

print(
    "ECLAT Frequent Itemsets:",
    len(eclat_df)
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print(
    "\nTop 10 Apriori Rules:"
)

print(
    rules_apriori[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(10)
)

print(
    "\nTop 10 FP-Growth Rules:"
)

print(
    rules_fp[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(10)
)

print(
    "\nResults saved successfully!"
)

print(
    "1. apriori_frequent_itemsets.csv"
)

print(
    "2. apriori_rules.csv"
)

print(
    "3. fpgrowth_frequent_itemsets.csv"
)

print(
    "4. fpgrowth_rules.csv"
)

print(
    "5. eclat_frequent_itemsets.csv"
)

print(
    "6. pca_results.csv"
)

print(
    "7. pca_visualization.png"
)

print(
    "8. top_15_items.png"
)
```
